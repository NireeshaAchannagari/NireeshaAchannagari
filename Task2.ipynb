```python
import random
import os

LOG_FILE = "guessing_game_log.txt"

def log_game_result(player_name, attempts, won):
    with open(LOG_FILE, "a") as file:
        status = "Won" if won else "Lost"
        file.write(f"{player_name},{status},{attempts} attempts\n")

def number_guessing_game():
    number_to_guess = random.randint(1, 100)
    max_attempts = 10
    attempts = 0

    print("🎯 Welcome to the Number Guessing Game!")
    print("I'm thinking of a number between 1 and 100.")
    print(f"You have {max_attempts} attempts to guess it.")

    player_name = input("Enter your name: ")

    while attempts < max_attempts:
        try:
            guess = int(input("Take a guess: "))
            attempts += 1

            if guess < 1 or guess > 100:
                print("Please enter a number between 1 and 100.")
            elif guess < number_to_guess:
                print("Too low!")
            elif guess > number_to_guess:
                print("Too high!")
            else:
                print(f"🎉 Congratulations, {player_name}! You guessed it in {attempts} attempts.")
                log_game_result(player_name, attempts, True)
                break
        except ValueError:
            print("Invalid input. Please enter a whole number.")

    else:
        print(f"😢 Sorry, {player_name}. You've used all {max_attempts} attempts.")
        print(f"The number was: {number_to_guess}")
        log_game_result(player_name, attempts, False)

def view_game_history():
    if not os.path.exists(LOG_FILE):
        print("No game history found.")
        return
    print("\n📜 Game History:")
    with open(LOG_FILE, "r") as file:
        for line in file:
            print(line.strip())

def main_menu():
    while True:
        print("\n--- Number Guessing Game Menu ---")
        print("1. Play Game")
        print("2. View Game History")
        print("3. Exit")

        choice = input("Enter your choice (1-3): ")

        if choice == "1":
            number_guessing_game()
        elif choice == "2":
            view_game_history()
        elif choice == "3":
            print("Goodbye! 👋")
            break
        else:
            print("Invalid choice. Please select 1, 2 or 3.")

main_menu()
```
