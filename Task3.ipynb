```python
import os

LOG_FILE = "temperature_conversion_log.txt"

def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

def celsius_to_kelvin(c):
    return c + 273.15

def fahrenheit_to_celsius(f):
    return (f - 32) * 5/9

def fahrenheit_to_kelvin(f):
    return (f - 32) * 5/9 + 273.15

def kelvin_to_celsius(k):
    return k - 273.15

def kelvin_to_fahrenheit(k):
    return (k - 273.15) * 9/5 + 32

def log_conversion(entry):
    with open(LOG_FILE, "a") as file:
        file.write(entry + "\n")

def view_conversion_history():
    if not os.path.exists(LOG_FILE):
        print("No conversion history found.")
        return
    print("\n📜 Conversion History:")
    with open(LOG_FILE, "r") as file:
        for line in file:
            print(line.strip())

def temperature_converter():
    while True:
        print("\n--- Temperature Converter ---")
        print("1. Celsius to Fahrenheit & Kelvin")
        print("2. Fahrenheit to Celsius & Kelvin")
        print("3. Kelvin to Celsius & Fahrenheit")
        print("4. View Conversion History")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ")

        if choice == "1":
            try:
                c = float(input("Enter temperature in Celsius: "))
                f = celsius_to_fahrenheit(c)
                k = celsius_to_kelvin(c)
                print(f"{c} °C = {f:.2f} °F = {k:.2f} K")
                log_conversion(f"{c} °C -> {f:.2f} °F, {k:.2f} K")
            except ValueError:
                print("Invalid input. Please enter a number.")
        elif choice == "2":
            try:
                f = float(input("Enter temperature in Fahrenheit: "))
                c = fahrenheit_to_celsius(f)
                k = fahrenheit_to_kelvin(f)
                print(f"{f} °F = {c:.2f} °C = {k:.2f} K")
                log_conversion(f"{f} °F -> {c:.2f} °C, {k:.2f} K")
            except ValueError:
                print("Invalid input. Please enter a number.")
        elif choice == "3":
            try:
                k = float(input("Enter temperature in Kelvin: "))
                c = kelvin_to_celsius(k)
                f = kelvin_to_fahrenheit(k)
                print(f"{k} K = {c:.2f} °C = {f:.2f} °F")
                log_conversion(f"{k} K -> {c:.2f} °C, {f:.2f} °F")
            except ValueError:
                print("Invalid input. Please enter a number.")
        elif choice == "4":
            view_conversion_history()
        elif choice == "5":
            print("Goodbye! 👋")
            break
        else:
            print("Invalid choice. Please select 1-5.")

temperature_converter()
```
